## Introduction to Visualization Libraries

Python has many visualization libraries for different purposes:

**Matplotlib** - The foundation library we'll use today:
- Most widely used, highly customizable
- Great for static plots, publications
- Steeper learning curve but very powerful

**Other libraries** (not covered today, but good to know):
- **Seaborn**: Built on matplotlib, beautiful statistical plots with less code
- **Plotly**: Interactive plots for dashboards and web apps
- **Bokeh**: Interactive visualizations for large datasets
- **Altair**: Declarative statistical visualization based on Vega-Lite

💡 Start with matplotlib to understand the basics, then explore others for specific needs!

## Matplotlib Basics

**1. Simple line plot** - Getting started with matplotlib

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

# Simple line plot
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.figure(figsize=(8, 5))
plt.plot(x, y, 'o-', linewidth=2, markersize=8)
plt.xlabel('X axis label', fontsize=12)
plt.ylabel('Y axis label', fontsize=12)
plt.title('This is the Title', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.show()

print("Key components:")
print("- Figure: The whole canvas")
print("- Title, labels, grid, etc.: All customizable")
```

**2. Subplots** - Multiple plots in one figure

```python
x = np.linspace(0, 10, 100)

plt.figure(figsize=(12, 8))

# Top left
plt.subplot(2, 2, 1)
plt.plot(x, np.sin(x))
plt.title('Sine')
plt.grid(True, alpha=0.3)

# Top right
plt.subplot(2, 2, 2)
plt.plot(x, np.cos(x), color='red')
plt.title('Cosine')
plt.grid(True, alpha=0.3)

# Bottom left
plt.subplot(2, 2, 3)
plt.plot(x, np.exp(-x/5), color='green')
plt.title('Exponential Decay')
plt.grid(True, alpha=0.3)

# Bottom right
plt.subplot(2, 2, 4)
plt.plot(x, x**2, color='purple')
plt.title('Quadratic')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

**3. Scatter plots with colors and sizes** - Encoding additional information

```python
# Generate data with categories
np.random.seed(42)
n = 100
x = np.random.randn(n)
y = np.random.randn(n)
colors = np.random.randn(n)  # Color values
sizes = np.random.randint(20, 200, n)  # Size values

plt.figure(figsize=(10, 6))
scatter = plt.scatter(x, y, c=colors, s=sizes, alpha=0.6, cmap='viridis')
plt.colorbar(scatter, label='Color value')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Scatter Plot with Colors and Sizes')
plt.grid(True, alpha=0.3)
plt.show()
```

**4. Histograms** - Visualizing distributions

```python
# Generate random normal data
np.random.seed(42)
data = np.random.randn(1000)

plt.figure(figsize=(10, 5))
plt.hist(data, bins=30, color='skyblue', edgecolor='black', alpha=0.7)
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram of Normal Distribution')
plt.grid(axis='y', alpha=0.3)
plt.show()

print(f"Mean: {data.mean():.2f}, Std: {data.std():.2f}")
```

**5. Box plots** - Visualizing quartiles and outliers

```python
np.random.seed(42)
data = [np.random.normal(0, std, 100) for std in range(1, 5)]

plt.figure(figsize=(10, 6))
plt.boxplot(data, tick_labels=['A', 'B', 'C', 'D'])
plt.ylabel('Value')
plt.title('Box Plot')
plt.grid(axis='y', alpha=0.3)
plt.show()

print("Box plot components:")
print("- Box: 25th to 75th percentile (IQR)")
print("- Line in box: Median (50th percentile)")
print("- Whiskers: ~1.5 * IQR")
print("- Points: Outliers beyond whiskers")
```

**6. Customizing colors** - Using color names, hex codes, and RGB

```python
x = np.arange(5)
y = [1, 2, 3, 4, 5]

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.bar(x, y, color='coral')  # Named color
plt.title('Named color: coral')

plt.subplot(1, 3, 2)
plt.bar(x, y, color='#FF6B6B')  # Hex code
plt.title('Hex code: #FF6B6B')

plt.subplot(1, 3, 3)
plt.bar(x, y, color=(0.2, 0.6, 0.8))  # RGB tuple (0-1 range)
plt.title('RGB: (0.2, 0.6, 0.8)')

plt.tight_layout()
plt.show()
```

**7. Adding annotations and text** - Highlighting important features

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.figure(figsize=(10, 6))
plt.plot(x, y, linewidth=2)

# Add annotation with arrow
plt.annotate('First peak', xy=(np.pi/2, 1), xytext=(2, 1.3),
             arrowprops=dict(arrowstyle='->', color='red', lw=2),
             fontsize=12, color='red')

# Add text without arrow
plt.text(8, -0.5, 'Important region', fontsize=12, 
         bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.5))

plt.xlabel('X')
plt.ylabel('Y')
plt.title('Plot with Annotations')
plt.grid(True, alpha=0.3)
plt.show()
```

## Why Visualization Matters

**9. Anscombe's Quartet** - Four datasets with identical statistics but very different patterns

```python
# Anscombe's quartet data
anscombe = pd.DataFrame({
    'x1': [10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5],
    'y1': [8.04, 6.95, 7.58, 8.81, 8.33, 9.96, 7.24, 4.26, 10.84, 4.82, 5.68],
    'x2': [10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5],
    'y2': [9.14, 8.14, 8.74, 8.77, 9.26, 8.10, 6.13, 3.10, 9.13, 7.26, 4.74],
    'x3': [10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5],
    'y3': [7.46, 6.77, 12.74, 7.11, 7.81, 8.84, 6.08, 5.39, 8.15, 6.42, 5.73],
    'x4': [8, 8, 8, 8, 8, 8, 8, 19, 8, 8, 8],
    'y4': [6.58, 5.76, 7.71, 8.84, 8.47, 7.04, 5.25, 12.50, 5.56, 7.91, 6.89]
})

# Calculate statistics for each dataset
for i in range(1, 5):
    x_col = f'x{i}'
    y_col = f'y{i}'
    print(f"Dataset {i}:")
    print(f"  Mean x: {anscombe[x_col].mean():.2f}, Mean y: {anscombe[y_col].mean():.2f}")
    print(f"  Std x: {anscombe[x_col].std():.2f}, Std y: {anscombe[y_col].std():.2f}")
    print(f"  Correlation: {anscombe[x_col].corr(anscombe[y_col]):.3f}")
    print()
```

**10. Visualizing Anscombe's Quartet** - The statistics are the same, but the data are completely different!

```python
plt.figure(figsize=(10, 10))

for i in range(4):
    x_col = f'x{i+1}'
    y_col = f'y{i+1}'
    
    plt.subplot(2, 2, i+1)
    plt.scatter(anscombe[x_col], anscombe[y_col], s=80)
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title(f'Dataset {i+1}')
    plt.xlim(0, 20)
    plt.ylim(0, 14)

plt.tight_layout()
plt.show()

print("\n💡 Lesson: ALWAYS visualize your data! Statistics alone can be misleading.")
```

## Importing and Visualizing Images

**11. Loading and displaying an image** - Images are just NumPy arrays!

```python
# Download a sample image
import urllib.request
url = 'https://picsum.photos/400/300'
urllib.request.urlretrieve(url, 'sample.jpg')

# Load using PIL and convert to numpy array
img_pil = Image.open('sample.jpg')
img = np.array(img_pil)

print(f"Image shape: {img.shape}")
print(f"Data type: {img.dtype}")
print(f"Value range: {img.min()} to {img.max()}")

plt.figure(figsize=(8, 6))
plt.imshow(img)
plt.axis('off')
plt.title('Color Image')
plt.show()

print(f"\n💡 RGB image: shape is (height, width, 3) - 3 channels for Red, Green, Blue")
print(f"   Images are NumPy arrays, so we can use all NumPy operations!")
```

**12. Extracting color channels** - Separating R, G, B using NumPy slicing

```python
# Extract individual channels using NumPy slicing
red_channel = img[:, :, 0]    # All rows, all columns, first channel
green_channel = img[:, :, 1]  # All rows, all columns, second channel
blue_channel = img[:, :, 2]   # All rows, all columns, third channel

plt.figure(figsize=(16, 4))

plt.subplot(1, 4, 1)
plt.imshow(img)
plt.title('Original')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(red_channel, cmap='Reds')
plt.title('Red Channel')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(green_channel, cmap='Greens')
plt.title('Green Channel')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(blue_channel, cmap='Blues')
plt.title('Blue Channel')
plt.axis('off')

plt.tight_layout()
plt.show()

print("💡 Channel extraction uses NumPy slicing: img[:, :, channel_index]")
```

**13. Cropping images** - Extracting regions using NumPy slicing

```python
# Crop the center region using NumPy slicing
height, width = img.shape[:2]
crop_size = min(height, width) // 2

center_y, center_x = height // 2, width // 2
y1, y2 = center_y - crop_size // 2, center_y + crop_size // 2
x1, x2 = center_x - crop_size // 2, center_x + crop_size // 2

# Crop using slicing: img[row_start:row_end, col_start:col_end]
cropped = img[y1:y2, x1:x2]

plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title('Original')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cropped)
plt.title(f'Cropped ({crop_size}x{crop_size})')
plt.axis('off')

plt.tight_layout()
plt.show()

print(f"Original shape: {img.shape}")
print(f"Cropped shape: {cropped.shape}")
print(f"\n💡 Cropping is just NumPy slicing: img[y1:y2, x1:x2]")
```

**14. Flipping and rotating images** - Using NumPy operations

```python
# All operations use NumPy array slicing and functions

# Flip vertically: reverse row order
flipped_v = img[::-1, :, :]  # or np.flipud(img)

# Flip horizontally: reverse column order
flipped_h = img[:, ::-1, :]  # or np.fliplr(img)

# Rotate 90 degrees counter-clockwise
rotated_90 = np.rot90(img)

# Rotate 180 degrees
rotated_180 = np.rot90(img, k=2)

plt.figure(figsize=(15, 10))

plt.subplot(2, 3, 1)
plt.imshow(img)
plt.title('Original')
plt.axis('off')

plt.subplot(2, 3, 2)
plt.imshow(flipped_v)
plt.title('Flipped Vertical\n(img[::-1, :, :])')
plt.axis('off')

plt.subplot(2, 3, 3)
plt.imshow(flipped_h)
plt.title('Flipped Horizontal\n(img[:, ::-1, :])')
plt.axis('off')

plt.subplot(2, 3, 4)
plt.imshow(rotated_90)
plt.title('Rotated 90°\n(np.rot90(img))')
plt.axis('off')

plt.subplot(2, 3, 5)
plt.imshow(rotated_180)
plt.title('Rotated 180°\n(np.rot90(img, k=2))')
plt.axis('off')

plt.tight_layout()
plt.show()

print("\n💡 All image manipulations use NumPy operations:")
print("   - Flipping: reverse slicing with [::-1]")
print("   - Rotating: np.rot90()")
print("   - Images are just arrays, so any NumPy operation works!")
```

**15. Loading MNIST digits** - Classic handwritten digit dataset

```python
# Load the MNIST dataset using fetch_openml
from sklearn.datasets import fetch_openml

mnist = fetch_openml('mnist_784', version=1, parser='auto')
X_mnist = mnist.data.values.reshape(-1, 28, 28)  # Reshape to 28x28 images
y_mnist = mnist.target.values.astype(int)

print(f"Dataset shape: {X_mnist.shape}")
print(f"Image shape: {X_mnist[0].shape}")
print(f"Data type: {X_mnist.dtype}")
print(f"Value range: {X_mnist.min():.0f} to {X_mnist.max():.0f}")
print(f"Labels: {np.unique(y_mnist)}")

print(f"\n💡 MNIST contains 28x28 grayscale images of handwritten digits (0-9)")
print(f"   70,000 images total (we typically use 60k for training, 10k for testing)")
```

## High-Dimensional Data Visualization

**17. The curse of dimensionality** - Why we need dimensionality reduction

```python
print("High-dimensional data challenges:")
print()
print("1. 📊 Visualization: Can't plot 100 dimensions!")
print("2. 💭 Intuition: Hard to understand relationships")
print("3. 💾 Computation: Many features = slow algorithms")
print("4. 🎯 Noise: More dimensions = more irrelevant features")
print()
print("Examples of high-dimensional data:")
print("- MNIST digits: 784 dimensions (28x28 pixels)")
print("- Neural recordings: hundreds of neurons × time points")
print("- Gene expression: thousands of genes per cell")
print()
print("💡 Solution: Dimensionality reduction to 2D or 3D for visualization!")
print()
print("Common techniques:")
print("- PCA (Principal Component Analysis): Linear, fast, interpretable")
print("- t-SNE (t-distributed Stochastic Neighbor Embedding): Non-linear, great for visualization")
print("- UMAP (Uniform Manifold Approximation and Projection): Like t-SNE but faster")
print()
print("Today we'll focus on t-SNE!")
```

**18. Loading Iris dataset** - Classic example for dimensionality reduction

```python
from sklearn.datasets import load_iris

# Load Iris dataset
iris = load_iris()
X = iris.data  # Features
y = iris.target  # Labels

print(f"Data shape: {X.shape}")
print(f"Features: {iris.feature_names}")
print(f"Classes: {iris.target_names}")
print(f"\nFirst 5 samples:")
print(X[:5])
```

**19. t-SNE (t-Distributed Stochastic Neighbor Embedding)** - Non-linear dimensionality reduction

```python
from sklearn.manifold import TSNE

# Apply t-SNE
tsne = TSNE(n_components=2, random_state=42, perplexity=30)
X_tsne = tsne.fit_transform(X)

print(f"Original shape: {X.shape}")
print(f"t-SNE shape: {X_tsne.shape}")
print("\n💡 t-SNE is non-linear and preserves local structure")
print("   It's great for finding clusters in high-dimensional data!")
```

**20. Visualizing t-SNE results** - Color coding by class and by features

```python
plt.figure(figsize=(16, 6))

# Left: Color by species (categorical)
plt.subplot(1, 2, 1)
colors = ['red', 'green', 'blue']
for i, species in enumerate(iris.target_names):
    mask = y == i
    plt.scatter(X_tsne[mask, 0], X_tsne[mask, 1], 
                c=colors[i], label=species, s=50, alpha=0.7)
plt.xlabel('t-SNE 1')
plt.ylabel('t-SNE 2')
plt.title('t-SNE colored by Species (categorical)')
plt.legend()
plt.grid(True, alpha=0.3)

# Right: Color by petal length (continuous)
plt.subplot(1, 2, 2)
feature_idx = 2  # Petal length
feature_name = iris.feature_names[feature_idx]
scatter = plt.scatter(X_tsne[:, 0], X_tsne[:, 1], 
                     c=X[:, feature_idx], cmap='viridis', s=50, alpha=0.7)
plt.colorbar(scatter, label=feature_name)
plt.xlabel('t-SNE 1')
plt.ylabel('t-SNE 2')
plt.title(f't-SNE colored by {feature_name} (continuous)')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("\n💡 t-SNE creates tight, well-separated clusters!")
print("   - Left: Categorical coloring shows species separation")
print("   - Right: Continuous coloring shows feature variation")
```

**21. t-SNE on MNIST digits** - Visualizing high-dimensional image data

```python
# Use a subset for faster computation
n_samples = 500
X_mnist_subset = mnist.data.values[:n_samples]  # Already flattened (784 dimensions)
y_mnist_subset = y_mnist[:n_samples]

print(f"Applying t-SNE to {n_samples} MNIST images...")
print(f"Original dimensionality: {X_mnist_subset.shape[1]} (28×28 flattened)")

tsne_mnist = TSNE(n_components=2, random_state=42, perplexity=30)
X_mnist_tsne = tsne_mnist.fit_transform(X_mnist_subset)

print(f"Reduced dimensionality: {X_mnist_tsne.shape[1]}")
print("Done!\n")

# Visualize
plt.figure(figsize=(12, 10))
scatter = plt.scatter(X_mnist_tsne[:, 0], X_mnist_tsne[:, 1], 
                     c=y_mnist_subset, cmap='tab10', s=30, alpha=0.7)
plt.colorbar(scatter, label='Digit', ticks=range(10))
plt.xlabel('t-SNE 1')
plt.ylabel('t-SNE 2')
plt.title('t-SNE of MNIST Digits')
plt.grid(True, alpha=0.3)
plt.show()

print("\n💡 Similar digits cluster together!")
print("   Notice how the model discovers structure without being told about digit identity")
print("   From 784 dimensions (28×28 pixels) down to 2D, while preserving neighborhoods!")
```

## Exercise 3: Custom Scatter Plot

Create a scatter plot where:
1. Generate 100 random (x, y) points from a normal distribution
2. Color each point based on its distance from the origin (use sqrt(x² + y²))
3. Make point size proportional to the same distance
4. Add a colorbar, labels, title, and grid
5. Use the 'coolwarm' colormap

### Solution 3

```python
# Solution 3: Custom Scatter Plot

np.random.seed(42)

# 1. Generate 100 random points
n = 100
x = np.random.randn(n)
y = np.random.randn(n)

# 2. Calculate distance from origin
distances = np.sqrt(x**2 + y**2)

# 3. Size proportional to distance
sizes = distances * 50  # Scale factor for visibility

# 4. Create scatter plot
plt.figure(figsize=(10, 8))
scatter = plt.scatter(x, y, c=distances, s=sizes, alpha=0.6, cmap='coolwarm')
plt.colorbar(scatter, label='Distance from Origin')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Scatter Plot: Color and Size by Distance from Origin')
plt.grid(True, alpha=0.3)

# Add origin marker
plt.scatter(0, 0, c='black', s=200, marker='x', linewidths=3, label='Origin')
plt.legend()

plt.show()

print(f"Distance range: {distances.min():.2f} to {distances.max():.2f}")
```
